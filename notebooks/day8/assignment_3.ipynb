# Assignment 3: Rule-Based Product Recommendation Engine

Is assignment mein hum ek **Rule-Based Recommendation System** (No Machine Learning) design aur implement karenge.

### 🎯 Business Logic & Rules:
1. **Rule 1 (Happy Customer - Rating $\ge$ 4.0)**:
   - Agar user ne 4 ya 4 se upar rating di hai, iska matlab use brand pasand aaya hai.
   - **Action**: Use **Same Brand** ke doosre products recommend karo (Brand Loyalty & Upselling).
2. **Rule 2 (Unhappy/Average Customer - Rating < 4.0)**:
   - Agar user ne 4 se kam rating di hai, iska matlab wo is brand se satisfied nahi hai.
   - **Action**: Use **Different Brands** (Competitor alternatives) ke products dikhao taaki customer churn na ho.

## Step 1: E-Commerce Product Catalog Banate Hain
Har e-commerce system mein ek product catalog hota hai jisme product name, brand, category aur price hoti hai.

```python
# E-Commerce Product Inventory / Catalog
product_catalog = [
    {"id": 101, "name": "Samsung Galaxy S24", "brand": "Samsung", "category": "Smartphone", "price": 79999},
    {"id": 102, "name": "Samsung Galaxy Watch 6", "brand": "Samsung", "category": "Smartwatch", "price": 29999},
    {"id": 103, "name": "Samsung Galaxy Buds 2", "brand": "Samsung", "category": "Audio", "price": 11999},
    
    {"id": 104, "name": "Apple iPhone 15", "brand": "Apple", "category": "Smartphone", "price": 74999},
    {"id": 105, "name": "Apple Watch Series 9", "brand": "Apple", "category": "Smartwatch", "price": 41999},
    {"id": 106, "name": "Apple AirPods Pro", "brand": "Apple", "category": "Audio", "price": 24999},
    
    {"id": 107, "name": "Sony WH-1000XM5", "brand": "Sony", "category": "Audio", "price": 29999},
    {"id": 108, "name": "Sony Bravia 55-inch 4K TV", "brand": "Sony", "category": "Television", "price": 64999},
    
    {"id": 109, "name": "OnePlus 12", "brand": "OnePlus", "category": "Smartphone", "price": 64999},
    {"id": 110, "name": "OnePlus Nord Buds 2", "brand": "OnePlus", "category": "Audio", "price": 2999}
]

print(f"Total Products in Catalog: {len(product_catalog)}")
all_brands = list(set(p['brand'] for p in product_catalog))
print("Available Brands:", all_brands)
```

## Step 2: Customer Purchase & Review Data
Assignment 1 ke cleaned customers ke saath dekhte hain unhone kis brand ka product review kiya tha.

```python
# Cleaned Customers from Assignment 1 with purchased brand
customer_reviews = [
    {"name": "Alice", "rating": 5.0, "brand": "Samsung", "purchased_item": "Samsung Galaxy S24"},
    {"name": "Bob",   "rating": 4.0, "brand": "Apple",   "purchased_item": "Apple iPhone 15"},
    {"name": "Diana", "rating": 5.0, "brand": "Sony",    "purchased_item": "Sony WH-1000XM5"},
    {"name": "Rahul", "rating": 2.0, "brand": "Samsung", "purchased_item": "Samsung Galaxy S24"},
    {"name": "Eve",   "rating": 3.5, "brand": "OnePlus", "purchased_item": "OnePlus 12"}
]

for c in customer_reviews:
    print(f"Customer: {c['name']:<6} | Rating: {c['rating']} | Brand: {c['brand']:<8} | Item: {c['purchased_item']}")
```

## Step 3: Recommendation Engine (Rule-Based Logic)
Ab hum system design ka core algorithm banayenge jo dono rules follow karta hai:
- Agar `rating >= 4.0` -> `Same Brand` ke doosre products filter karo (jo usne already buy nahi kiya).
- Agar `rating < 4.0` -> `Different Brands` ke top products recommend karo.

```python
def recommend_products(customer_name, rating, current_brand, purchased_item, catalog):
    recommendations = []
    
    if rating >= 4.0:
        strategy = "SAME BRAND (Customer is Satisfied)"
        # Rule 1: Same brand products, but exclude current purchased item
        recommendations = [
            p for p in catalog 
            if p["brand"].lower() == current_brand.lower() and p["name"] != purchased_item
        ]
    else:
        strategy = "DIFFERENT BRAND (Customer is Dissatisfied - Showing Competitor Alternatives)"
        # Rule 2: Competitor brands products
        recommendations = [
            p for p in catalog 
            if p["brand"].lower() != current_brand.lower()
        ]
        
    return strategy, recommendations

print("Recommendation Engine Function Ready!")
```

## Step 4: Sabhi Customers Par Test Run Karte Hain
Aayiye dekhte hain Alice, Bob, Diana, Rahul aur Eve ko kya-kya recommendations milti hain!

```python
print("=" * 75)
print("       RULE-BASED RECOMMENDATION ENGINE OUTPUT")
print("=" * 75)

for cust in customer_reviews:
    strategy, recs = recommend_products(
        customer_name=cust["name"],
        rating=cust["rating"],
        current_brand=cust["brand"],
        purchased_item=cust["purchased_item"],
        catalog=product_catalog
    )
    
    print(f"\n👤 Customer: {cust['name']} (Rating: {cust['rating']} ⭐ for {cust['brand']})")
    print(f"   Strategy: {strategy}")
    print(f"   Recommended Products ({len(recs)}): ")
    for r in recs[:3]:  # Top 3 recommendations show karte hain
        print(f"     -> [{r['brand']}] {r['name']} (Rs. {r['price']})")

print("\n" + "=" * 75)
```

## Step 5: Interactive Custom Tester
Aap yahan custom input dekar test kar sakte hain!

```python
# Example: Ek naye customer ka test
test_user = "Karan"
test_rating = 1.5  # Poor rating
test_brand = "Apple"
test_item = "Apple iPhone 15"

strat, results = recommend_products(test_user, test_rating, test_brand, test_item, product_catalog)
print(f"Test User: {test_user} | Rating: {test_rating} ⭐ | Brand: {test_brand}")
print(f"Strategy Applied: {strat}")
print("Recommended Alternative Brands:")
for prod in results[:3]:
    print(f"  * {prod['name']} ({prod['brand']}) - Rs. {prod['price']}")
```
