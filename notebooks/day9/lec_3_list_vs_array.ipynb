# Lecture 3: Python List vs NumPy Array (Performance Benchmark)

NumPy Python List se do fundamental architectural reasons ki wajah se kayi guna tez aur memory-efficient hota hai:
1. **Vectorization**: Python ke slow `for` loops ke bajaye C-level hardware SIMD instructions use karta hai.
2. **Contiguous Memory**: Numbers memory (RAM) mein bagal-bagal continuous blocks mein store hote hain, jisme pointer overhead bilkul zero hota hai.

---

## Topic 1: Execution Speed (Time Benchmark)

Hum 1 Crore (10,000,000) numbers par har element ko 2 se multiply (`x * 2`) karke dono ka execution time measure karenge.

### Step 1.1: Python List Test
- Python list comprehension internally ek Python-level for-loop chalati hai, jisme har iteration par dynamic type checking ka overhead hota hai.

```python
import time

size = 10_000_000
py_list = list(range(size))

start_time = time.time()
py_result = [x * 2 for x in py_list]
py_duration = time.time() - start_time

print(f"Python List Duration: {py_duration:.4f} seconds")
```

### Step 1.2: NumPy Array Test (Vectorized Operation)
- NumPy mein hume koi loop nahi likhna padta; `np_arr * 2` seedhe C-kernel level par parallel execute hota hai.

```python
import numpy as np

np_arr = np.arange(size)

start_time = time.time()
np_result = np_arr * 2  # Vectorized operation
np_duration = time.time() - start_time

print(f"NumPy Array Duration: {np_duration:.4f} seconds")
```

### Step 1.3: Speed Ratio Comparison
- Dekhte hain NumPy kitne guna fast execute hua.

```python
speedup = py_duration / np_duration
print(f"Result: NumPy is {speedup:.1f}x faster than Python List!")
```

## Topic 2: Memory Consumption (Bytes Benchmark)

- **Python List**: Har integer ek full Python Object hota hai jisme type info, reference count aur memory address store hoti hai (approx 28 bytes per integer).
- **NumPy Array**: Raw binary numbers (4 bytes ya 8 bytes) ek contiguous memory block mein bina kisi wrapper ke store hote hain.

### Step 2.1: Python List Memory Calculation (`sys.getsizeof`)
- 1,000 numbers ke liye Python list kitni RAM leti hai:

```python
import sys

sample_size = 1000
py_list_sample = list(range(sample_size))

# Ek integer ka size * total elements
py_list_bytes = sys.getsizeof(1) * len(py_list_sample)
print(f"Python List Memory : {py_list_bytes} bytes")
```

### Step 2.2: NumPy Array Memory Calculation (`arr.nbytes`)
- NumPy array ka `.nbytes` attribute array ke raw buffer dwara consume kiye gaye exact bytes batata hai:

```python
np_arr_sample = np.arange(sample_size)
np_arr_bytes = np_arr_sample.nbytes

print(f"NumPy Array Memory : {np_arr_bytes} bytes")
```

### Step 2.3: Memory Efficiency Ratio
- Dekhte hain Python list ne NumPy array ke mukable kitni guna zyada memory use ki:

```python
memory_ratio = py_list_bytes / np_arr_bytes
print(f"Result: Python List consumes {memory_ratio:.1f}x more memory than NumPy Array!")
```
