# Lecture 7: Reshaping Arrays & Flattening

NumPy mein **Reshaping** ka matlab hota hai data ko bina badle array ke dimensions (rows aur columns) ka structure change karna.

### 📌 The Golden Rule of Reshaping:
Reshape tabhi possible hota hai jab initial total elements aur reshaped total elements bilkul barabar hon:
$$\text{Old Rows} \times \text{Old Cols} = \text{New Rows} \times \text{New Cols}$$
Jaise agar array mein 6 elements hain, toh use `(2, 3)`, `(3, 2)`, `(1, 6)`, ya `(6, 1)` mein reshape kiya ja sakta hai. Agar size match nahi hoga toh NumPy `ValueError` dega.

---

## Step 1: Base 2D Array Banana (Shape: 2 x 3)
- Hum 2 rows aur 3 columns ka ek array banate hain jisme 1 se lekar 6 tak numbers hain ($2 \times 3 = 6$ elements):

```python
import numpy as np

# 2 rows aur 3 columns ka array
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Original Array:")
print(arr)
print("\nOriginal Shape :", arr.shape)  # (2, 3)
print("Total Elements :", arr.size)   # 6
```

---
## Step 2: `arr.reshape((3, 2))` (Rows & Columns Change Karna)
- **Concept**: Ab hum usi 6 elements wale array ko 3 rows aur 2 columns mein restructure karte hain.
- **Rule Check**: $2 \times 3 = 6$ tha, aur $3 \times 2 = 6$ hai. Elements ki count preserve rehti hai.

```python
# (2, 3) ko (3, 2) mein convert karte hain
reshaped = arr.reshape((3, 2))

print("Reshaped Array (3 Rows, 2 Columns):")
print(reshaped)
print("\nNew Shape :", reshaped.shape)  # (3, 2)
```

---
## Step 3: `arr.flatten()` (2D Matrix se 1D Vector Banana)
- **Concept**: Multi-dimensional array (2D ya 3D) ko wapas 1-dimensional flat array (vector) mein convert karne ke liye `.flatten()` use hota hai.
- **Machine Learning Use Case**: Computer Vision / Deep Learning mein 2D images (pixels grid) ko fully connected neural network layers mein input dene se pehle 1D array mein flatten kiya jata hai.
- **Result Shape**: `(6,)` (1D array with 6 elements).

```python
# 2D matrix ko 1D flat array banana
flattened = arr.flatten()

print("Flattened Array (1D Vector):")
print(flattened)
print("\nFlattened Shape   :", flattened.shape)  # (6,)
print("Dimensions (ndim) :", flattened.ndim)   # 1
```
