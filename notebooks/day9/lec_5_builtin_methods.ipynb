# Lecture 5: Creating Arrays using Built-in Methods

Manual list likhne ke bajaye NumPy mein frequently use hone wale standard arrays ko fast generate karne ke liye 6 powerful built-in methods hote hain.

---

## Method 1: `np.zeros(shape, dtype)`
- **Purpose**: Saare elements ko `0` se pre-fill karta hai.
- **Default Behavior**: Agar dtype na diya jaye toh yeh `float64` (`0.`) banata hai.
- **Integer Array**: Agar integers chahiye toh explicitly `dtype="int64"` pass karte hain.

```python
import numpy as np

# 2 rows aur 3 columns ka integer zero matrix
arr1 = np.zeros((2, 3), dtype="int64")
print("np.zeros Array:")
print(arr1)
print("Shape:", arr1.shape, "| Data Type:", arr1.dtype)
```

---
## Method 2: `np.ones(shape)`
- **Purpose**: Saare elements ko `1.` se pre-fill karta hai.
- By default iska data type `float64` hota hai.

```python
# 5 elements ka 1D ones array
arr2 = np.ones((5, ))
print("np.ones Array:")
print(arr2)
print("Shape:", arr2.shape, "| Data Type:", arr2.dtype)
```

---
## Method 3: `np.full(shape, fill_value)`
- **Purpose**: Kisi bhi custom constant number (jaise `100`, `-1`, ya `99`) se poora matrix fill karta hai.
- Pehla argument shape tuple hota hai `(rows, cols)` aur doosra argument fill value hoti hai.

```python
# 3 rows aur 4 columns ka matrix jisme har jagah 100 hoga
arr3 = np.full((3, 4), 100)
print("np.full Array:")
print(arr3)
print("Shape:", arr3.shape)
```

---
## Method 4: `np.eye(N)` (Identity Matrix)
- **Purpose**: Ek $N \times N$ **Identity Matrix** banata hai.
- **Structure**: Iske main diagonal (top-left se bottom-right) par `1.` aur baaki sabhi positions par `0.` hota hai.
- Linear Algebra aur Matrix multiplication mein Identity matrix ka critical role hota hai ($A \times I = A$).

```python
# 3x3 Identity Matrix
arr4 = np.eye(3)
print("np.eye (Identity Matrix):")
print(arr4)
print("Shape:", arr4.shape)
```

---
## Method 5: `np.arange(start, stop, step)`
- **Purpose**: Python ke standard `range()` ki tarah fixed jump ke sath sequential numbers generate karta hai.
- **Rule**: `stop` value **exclude** hoti hai (yaani `stop - 1` tak hi chalega).

```python
# 0 se shuru karke 11 tak, 2 ke step se (0, 2, 4, 6, 8, 10)
arr5 = np.arange(0, 11, 2)
print("np.arange Array:")
print(arr5)
print("Shape:", arr5.shape)
```

---
## Method 6: `np.linspace(start, stop, num)` (Linear Space)
- **Purpose**: Start aur Stop ke beech mathematically barabar faasle (evenly spaced) par `num` total points divide karta hai.
- **Crucial Difference from arange**: Yahan `stop` value **include** hoti hai (1 aur 100 dono result mein aayenge).

```python
# 1 se 100 ke beech barabar faasle par exactly 3 points (1.0, 50.5, 100.0)
arr6 = np.linspace(1, 100, 3)
print("np.linspace Array:")
print(arr6)
print("Shape:", arr6.shape)
```
