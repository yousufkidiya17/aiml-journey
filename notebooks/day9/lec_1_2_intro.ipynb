# Lecture 1 & 2: Introduction to NumPy & First Array

NumPy (**Numerical Python**) scientific computing, machine learning aur artificial intelligence ki foundation library hai. Python lists ke muqable yeh C-level optimization aur memory efficiency provide karti hai.

---

## Step 1: NumPy Library Import Karna
- Industry standard ke mutabiq NumPy ko hamesha alias `np` ke roop mein import kiya jata hai taaki har baar poora naam na likhna pade.

```python
import numpy as np

print("NumPy Successfully Imported! Version:", np.__version__)
```

## Step 2: 1D NumPy Array Banana (`np.array`)
- `np.array()` function kisi bhi standard Python list ko NumPy ke core data structure **`ndarray` (N-Dimensional Array)** mein convert kar deta hai.

```python
# Python list se 1D array banate hain
arr = np.array([1, 2, 3, 4, 5])
print("Array Content:", arr)
```

## Step 3: Object Type Check Karna (`type()`)
- Yeh verify karta hai ki Python object ka type kya hai. Python list ke bajaye yeh `<class 'numpy.ndarray'>` return karega.

```python
print("Object Structure Type:", type(arr))
```

## Step 4: Array Data Type (`dtype`) aur Shape (`shape`)
- **`arr.dtype`**: Array ke andar baithe elements ka binary data type (e.g., `int32` ya `int64`).
- **`arr.shape`**: Array ke dimensions ka tuple. `(5,)` ka matlab hai yeh ek 1-dimensional array hai jisme 5 elements hain.

```python
print("Data Type (dtype) :", arr.dtype)
print("Array Shape       :", arr.shape)
```
